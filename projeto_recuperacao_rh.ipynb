```python
# ============================================================
# BLOCO 1: CARGA DOS DADOS E ESTATÍSTICAS SALARIAIS (TAREFA 2)
# ============================================================
import pandas as pd
import numpy as np

# 1. Carregar os 3 arquivos CSV exportados do FreeSQL
df_salarios = pd.read_csv('query_01.csv')
df_geografia = pd.read_csv('query_02.csv')
df_admissao = pd.read_csv('query_03.csv')

# Padronizar nomes das colunas para letras maiúsculas por garantia
df_salarios.columns = df_salarios.columns.str.upper()
df_geografia.columns = df_geografia.columns.str.upper()
df_admissao.columns = df_admissao.columns.str.upper()

# 2. Resumo das Estatísticas Descritivas dos Salários
salarios = df_salarios['SALARIO']

print("--- ESTATÍSTICAS SALARIAIS DOS FUNCIONÁRIOS ---")
print(f"Total de Funcionários Analisados: {salarios.count()}")
print(f"Média Salarial: R$ {salarios.mean():,.2f}")
print(f"Mediana Salarial: R$ {salarios.median():,.2f}")
print(f"Salário Mínimo: R$ {salarios.min():,.2f}")
print(f"Salário Máximo: R$ {salarios.max():,.2f}")
print(f"Desvio Padrão: R$ {salarios.std():,.2f}")
```

```python
# ============================================================
# BLOCO 2: GERAÇÃO DOS 2 GRÁFICOS (TAREFA 2 - RH)
# ============================================================
import matplotlib.pyplot as plt

# Criar a área com 2 gráficos lado a lado
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# ------------------------------------------------------------
# GRÁFICO 1: Histograma da Distribuição Salarial
# ------------------------------------------------------------
ax1.hist(df_salarios['SALARIO'], bins=8, color='#2b5c8f', edgecolor='black', alpha=0.8)
ax1.set_title('Distribuição de Salários dos Funcionários', fontsize=12, fontweight='bold')
ax1.set_xlabel('Faixa Salarial (R$)', fontsize=10)
ax1.set_ylabel('Quantidade de Funcionários', fontsize=10)
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# ------------------------------------------------------------
# GRÁFICO 2: Média Salarial por Ano de Admissão
# ------------------------------------------------------------
df_ano = df_admissao.groupby('ANO_ADMISSAO')['SALARIO'].mean().reset_index()
ax2.bar(df_ano['ANO_ADMISSAO'].astype(str), df_ano['SALARIO'], color='#e06d53', edgecolor='black', alpha=0.85)
ax2.set_title('Média Salarial por Ano de Admissão', fontsize=12, fontweight='bold')
ax2.set_xlabel('Ano de Contratação', fontsize=10)
ax2.set_ylabel('Média Salarial (R$)', fontsize=10)
ax2.tick_params(axis='x', rotation=45)
ax2.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
